```python
import pandas as pd

# Cargar el archivo
df = pd.read_csv("../data/data.csv")

# Mostrar las primeras filas del DataFrame
df.head(20)
```

```python
# Estado de los datos

df.info()
```

```python
# Resumen estadístico de las columnas numéricas
df.describe()
```

```python
promedio_por_fecha_escuderia = (
    df.groupby(["fecha", "escuderia"], as_index=False)["tiempo_carrera_min"]
      .mean()
      .rename(columns={"tiempo_carrera_min": "avg_tiempo_carrera_min"})
      .sort_values(["fecha", "avg_tiempo_carrera_min"], ascending=[True, True])
)

promedio_por_fecha_escuderia.head(15)
```
